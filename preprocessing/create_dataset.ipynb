# Image import and augmentation

```python
files = os.listdir(dir)
print(len(files))
img_num = len(files)
T = 500

for i in range(img_num // T):
    images = []
    batch_files = files[i * T:(i + 1) * T]

    for fname in batch_files:
        file = f"{dir}/{fname}"
        try: 
            img = cv.cvtColor(cv.imread(file), code=cv.COLOR_BGR2LAB)
            img = functions.autoCrop(img)
            images.append(img)
        except: pass
        
    images = np.stack(images) / 255
    masks = functions.extractColorMask(images)

    np.save(f"../dataset/labs/{i}.npy", images)
    np.save(f"../dataset/masks/{i}.npy", masks)
    
```
